```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.

import pandas as pd

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

# Number 1: how weak is the naive "search volume predicts traffic" assumption?
corr = df["search_volume"].corr(df["impressions_90d"])
print(f"1. Correlation search_volume vs impressions_90d: {corr:.3f}  (near zero -> volume alone doesn't predict traffic)")

# Number 2: how big is the CTR gap across position tiers, for pages that are actually visible?
visible = df[df["impressions_90d"] >= 100]
ctr_by_pos = visible.groupby("position_tier")["ctr"].mean().sort_values(ascending=False)
print(f"\n2. Mean CTR by position tier (impressions_90d >= 100):\n{ctr_by_pos.round(4).to_string()}")

# Number 3: what share of pages are both stale AND declining -- i.e. plausible refresh candidates?
stale = df["days_since_last_update"] >= 180
declining = df["trend_direction"].str.lower().eq("down")
share = (stale & declining).mean()
print(f"\n3. Share of pages that are stale (>=180 days) AND declining: {share:.1%}")
```

*Write what your work will be able to say (observed, directional, decision-support) — and what it never will (causal proof, 'predicting Google').*

**What this work can say:** observed correlations and patterns in past, anonymized performance data e.g. "pages with X characteristics were, in this dataset, more likely to be tagged declining" or "a model trained on these signals ranked recovery candidates better than a hand-written rule, measured by Precision@K on held-out pages."
